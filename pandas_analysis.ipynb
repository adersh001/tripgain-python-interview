```python
#Data Cleaning
df['city'] =df['city'].fillna('Unknown')
df['winner'] =df['winner'].fillna('Unknown')
df['player_of_match'] = df['player_of_match'].fillna('Unknown')
df['umpire1'] = df['umpire1'].fillna('Unknown')
df['umpire2'] = df['umpire2'].fillna('Unknown')
df.drop('umpire3',axis=1,inplace=True)
```

```python
# Which player has won the most “Player of the Match” awards in games decided on the final ball?

matches=df[(df['win_by_runs']==1) | (df['win_by_wickets']==1)]
player_of_the_match = matches['player_of_match'].value_counts().idxmax()
print(f"Player of the Match: {player_of_the_match}")
```

```python
# At Wankhede Stadium, is it more common to win by batting first (runs) or by batting second

stadium= df[df['venue']=='Wankhede Stadium']

stadium['winner'].value_counts()
```

```python
# Which team has the highest number of wins where the victory margin was greater than 50 runs?
win_greater_50 = df[df['win_by_runs']>50]
team = win_greater_50['winner'].value_counts().idxmax()

print(f"Team has the highest number of wins where the victory margin was greater than 50 runs: {team}")
```

```python
# Which of the two umpires (umpire1 or umpire2) has officiated more matches involving the Kolkata Knight Riders?
match_kkr = df[(df['team1'] == 'Kolkata Knight Riders') | (df['team2'] == 'Kolkata Knight Riders')]
umpire1 = match_kkr['umpire1'].value_counts().idxmax()
umpire2 = match_kkr['umpire2'].value_counts().idxmax()
print(f"Umpires officiated more matches involving the Kolkata Knight Riders are:")
print(umpire1)
print(umpire2)
```
